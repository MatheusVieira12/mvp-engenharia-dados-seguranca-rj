```python
from pyspark.sql import functions as F
```

```python
caminho_ocorrencias = "/Volumes/projeto_seguranca_rj/bronze/arquivos/BaseMunicipioMensal.csv"
caminho_taxas = "/Volumes/projeto_seguranca_rj/bronze/arquivos/BaseMunicipioTaxaMes.csv"
caminho_dp = "/Volumes/projeto_seguranca_rj/bronze/arquivos/BaseDPEvolucaoMensalCisp.csv"
```

```python
df_ocorrencias = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("sep", ";")
    .option("encoding", "latin1")
    .csv(caminho_ocorrencias)
)

print("Linhas da base de ocorrências:", df_ocorrencias.count())

display(df_ocorrencias.limit(10))
```

```python
df_ocorrencias.printSchema()
```

```python
df_ocorrencias_bronze = (
    df_ocorrencias
    .withColumn("fonte_arquivo", F.lit("BaseMunicipioMensal.csv"))
    .withColumn("data_ingestao", F.current_timestamp())
)
```

```python
df_taxas = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .option("sep", ";")
    .option("encoding", "latin1")
    .csv(caminho_taxas)
)

print("Linhas da base de taxas:", df_taxas.count())
display(df_taxas.limit(10))
```

```python
df_taxas.printSchema()
```

```python
df_taxas_bronze = (
    df_taxas
    .withColumn("fonte_arquivo", F.lit("BaseMunicipioTaxaMes.csv"))
    .withColumn("data_ingestao", F.current_timestamp())
)
```

```python
# ==========================================
# LEITURA BASE DP
# ==========================================


df_dp = (spark.read.option("header", True)
    .option("inferSchema", True)
    .option("sep", ";")
    .option("encoding", "latin1")
    .csv(caminho_dp)
)


print(
    "Linhas da base DP:",
    df_dp.count()
)


display(
    df_dp.limit(10)
)
```

```python
df_dp.printSchema()
```

```python
df_dp_bronze = (
    df_dp
    .withColumn("fonte_arquivo", F.lit("BaseDPEvolucaoMensalCisp.csv"))
    .withColumn("data_ingestao", F.current_timestamp())
)
```

```python
df_ocorrencias_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("projeto_seguranca_rj.bronze.ocorrencias_municipio")

df_taxas_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("projeto_seguranca_rj.bronze.taxas_municipio")   

df_dp_bronze.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "projeto_seguranca_rj.bronze.dp_municipio"
    )
```

```python
validacoes = [
    (
        "Ocorrências",
        df_ocorrencias_bronze,
        "projeto_seguranca_rj.bronze.ocorrencias_municipio"
    ),
    (
        "Taxas",
        df_taxas_bronze,
        "projeto_seguranca_rj.bronze.taxas_municipio"
    ),
    (
        "DP",
        df_dp_bronze,
        "projeto_seguranca_rj.bronze.dp_municipio"
    )
]
```

```python
for nome, df_origem, tabela in validacoes:
    linhas_origem = df_origem.count()
    linhas_bronze = spark.table(tabela).count()

    print(
        f"{nome}: origem={linhas_origem} | "
        f"Bronze={linhas_bronze}"
    )

    if linhas_origem != linhas_bronze:
        raise ValueError(
            f"Divergência na carga de {nome}"
        )
```
